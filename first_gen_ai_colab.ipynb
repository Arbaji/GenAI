```python
import pandas as pd
import os
os.chdir('/content/gdrive/MyDrive/Gen-AI')
data = pd.read_csv('robot_mission.csv')
rounds = data['round'].tolist()
numbers = data['number'].tolist()
positions = []
current_position = 0
for i in rounds:
  if i % 2 == 1:
    current_position = current_position - numbers[i-1]
  else:
    current_position = current_position + numbers[i-1]
  positions.append(current_position)
print("The robot currently stands at block", current_position)
print("It visited the following blocks", positions)
```

```python
import matplotlib.pyplot as plt
plt.xlabel('Round')
plt.ylabel('Position')
plt.title('Robot Position Over Rounds')
plt.grid()
plt.plot(rounds, positions, marker='x')
plt.axhline(y=0)
plt.xticks(rounds)
plt.show()
```
